## Merge Debt Service + Debt-to-Income

````python
dashboard_canada = debt_service_wide.merge(
    debt_income_wide,
    on=["date", "geography"],
    how="outer"
)

dashboard_canada = dashboard_canada.sort_values(["geography", "date"])

dashboard_canada.head()
````

````python
dashboard_canada.columns.tolist()
````

````python
dashboard_canada["disposable_income_growth_yoy"] = dashboard_canada.groupby(
    "geography"
)["disposable_income"].pct_change(4) * 100

dashboard_canada["debt_to_income_change_yoy"] = dashboard_canada.groupby(
    "geography"
)["debt_to_disposable_income"].diff(4)

dashboard_canada["debt_service_ratio_change_yoy"] = dashboard_canada.groupby(
    "geography"
)["debt_service_ratio"].diff(4)

dashboard_canada["interest_cost_change_yoy"] = dashboard_canada.groupby(
    "geography"
)["effective_interest_cost_total_loans"].diff(4)
````

````python
dashboard_canada.head()
````

````python
def score_debt_service_ratio(x):
    if pd.isna(x):
        return np.nan
    if x < 13:
        return 20
    elif x < 14:
        return 45
    elif x < 15:
        return 65
    else:
        return 85


def score_debt_to_income(x):
    if pd.isna(x):
        return np.nan
    if x < 160:
        return 20
    elif x < 175:
        return 45
    elif x < 185:
        return 65
    else:
        return 85


def score_interest_pressure(x):
    if pd.isna(x):
        return np.nan
    if x < 0:
        return 20
    elif x < 0.25:
        return 45
    elif x < 0.75:
        return 65
    else:
        return 85


def score_income_pressure(x):
    if pd.isna(x):
        return np.nan
    if x > 5:
        return 20
    elif x > 2:
        return 45
    elif x > 0:
        return 65
    else:
        return 85
````

````python
dashboard_canada["dsr_score"] = dashboard_canada["debt_service_ratio"].apply(
    score_debt_service_ratio
)

dashboard_canada["debt_income_score"] = dashboard_canada["debt_to_disposable_income"].apply(
    score_debt_to_income
)

dashboard_canada["interest_pressure_score"] = dashboard_canada["interest_cost_change_yoy"].apply(
    score_interest_pressure
)

dashboard_canada["income_pressure_score"] = dashboard_canada["disposable_income_growth_yoy"].apply(
    score_income_pressure
)
````

## Risk Score Weighting Methodology

The risk score uses **analyst-defined heuristic weights**. These are not official regulatory, Bank of Canada, or Statistics Canada weights. They are created for this project to turn multiple household credit-risk indicators into one simple monitoring score.

The purpose of the score is not to predict individual borrower default. The purpose is to help a risk analyst or risk manager quickly understand whether household credit conditions are becoming more stressed over time.

```python
risk_score =
    debt_service_ratio_score * 0.35 +
    debt_income_score * 0.30 +
    interest_pressure_score * 0.20 +
    income_pressure_score * 0.15

````python
dashboard_canada["risk_score"] = (
    dashboard_canada["dsr_score"] * 0.35 +
    dashboard_canada["debt_income_score"] * 0.30 +
    dashboard_canada["interest_pressure_score"] * 0.20 +
    dashboard_canada["income_pressure_score"] * 0.15
)
````

````python
def assign_risk_category(score):
    if pd.isna(score):
        return "Insufficient Data"
    if score < 40:
        return "Low Risk"
    elif score < 60:
        return "Moderate Risk"
    elif score < 75:
        return "Elevated Risk"
    else:
        return "High Risk"


dashboard_canada["risk_category"] = dashboard_canada["risk_score"].apply(
    assign_risk_category
)
````

````python
def assign_suggested_action(category):
    if category == "Low Risk":
        return "Continue standard monitoring"
    elif category == "Moderate Risk":
        return "Increase monitoring of household credit stress indicators"
    elif category == "Elevated Risk":
        return "Review vulnerable borrower segments and stress-test portfolios"
    elif category == "High Risk":
        return "Escalate to risk committee and review credit policy thresholds"
    else:
        return "Data review required"


dashboard_canada["suggested_action"] = dashboard_canada["risk_category"].apply(
    assign_suggested_action
)
````

## Fixing date format

````python
import re
from datetime import datetime

month_map = {
    "jan": "01",
    "feb": "02",
    "fév": "02",
    "mar": "03",
    "apr": "04",
    "avr": "04",
    "may": "05",
    "mai": "05",
    "jun": "06",
    "juin": "06",
    "jul": "07",
    "juil": "07",
    "juill": "07",
    "aug": "08",
    "août": "08",
    "sep": "09",
    "sept": "09",
    "oct": "10",
    "nov": "11",
    "dec": "12",
    "déc": "12"
}

def parse_insolvency_date(x):
    if isinstance(x, pd.Timestamp):
        return x.to_period("M").to_timestamp()
    
    if isinstance(x, datetime):
        return pd.to_datetime(x).to_period("M").to_timestamp()
    
    text = str(x).strip().lower()
    
    parsed = pd.to_datetime(text, errors="coerce")
    if pd.notna(parsed):
        return parsed.to_period("M").to_timestamp()
    
    year_match = re.search(r"(19|20)\d{2}", text)
    if not year_match:
        return pd.NaT
    
    year = year_match.group(0)
    
    for key, month_num in month_map.items():
        if key in text:
            return pd.to_datetime(f"{year}-{month_num}-01")
    
    return pd.NaT
````

````python
insolvency_long["date"] = insolvency_long["date"].apply(parse_insolvency_date)
````
